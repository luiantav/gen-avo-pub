```python
%load_ext autoreload
%autoreload 2

from gp_functions import *
import jax
import jax.numpy as jnp
import numpy as np
import pandas as pd
import requests
import os
import pickle
import argparse

import matplotlib.pyplot as plt
from sklearn.preprocessing import minmax_scale
from scipy.stats import multivariate_normal
import matplotlib.colors as mcolors

import numpyro.distributions as dist
import seaborn as sns
import math
import itertools

import arviz as az
```

### Setup

```python
GRID_SIZE = 11
cs = 60 #cs position idx
val = 80 #cs value
row, col = divmod(cs, 11) #position coordinates
cs_position = [row,col]

N_SUBS = 1
N_BLOCKS = 1
N_TRIALS = 1 #in this case cs+ shown once 
y = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))#3D

#get euc distances to sample at different distances from CS+
distance_euc = np.zeros((GRID_SIZE, GRID_SIZE))
for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        distance_euc[i, j] = np.sqrt((i - cs_position[0]) ** 2 + (j - cs_position[1]) ** 2)
```

### Create all possible sampling combinations

```python
all_combinations = [list(comb) for r in range(2, 6) for comb in itertools.combinations(range(1, 6), r)]
opts = list(map(list, set(map(tuple, all_combinations))))
opts

#len(opts)

#for now subset some to make sure recovery works in the first place 
opts = [[1,3,4,5],
 [1,2,4]]
opts
```

### Simulate some RBF/LIN data

```python
# Make all possible coordinates
x1 = np.arange(GRID_SIZE)[:, None]
x2 = np.arange(GRID_SIZE)[:, None]
X = jnp.dstack(np.meshgrid(x1.squeeze(), x2.squeeze())).reshape(-1, 2)

# precompute distances between values in X
distances = jnp.sqrt(square_dist(X, 1)) 

y[0,0,0] = val #cs value
y_flat = np.reshape(y, (N_SUBS * N_BLOCKS, N_TRIALS))

mean = 0
lss = np.arange(0.5, 5.5, 0.5) #range ls parameter
grids = np.zeros((len(lss)*2, 121))

for cs in [cs]: # example locations on grid
    for i in range(len(lss)):
        ls = lss[i]
        cell = np.array(cs).reshape(1, 1)
        cell.astype(np.int32)
        
        #simulate some rbf data at different ls
        mu, var = gp_predict(cell, y_flat, distances, ls, mean)
        mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
        mu_flat = mu.squeeze()
        grids[i, :] = mu_flat #collect all grids
    
    
    for i in range(len(lss)):
        ls = lss[i]
        cell = np.array(cs).reshape(1, 1)
        cell.astype(np.int32)
        
        #simulate some lin data at different ls (and save at i+len subs)
        mu, var = gp_predict_lin(cell, y_flat, distances, ls, mean)
        mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
        mu_flat = mu.squeeze()
        grids[i+len(lss), :] = mu_flat #collect all grids
```

### Plot to make sure we created right thing

```python
n_plots = grids.shape[0]  
fig, axes = plt.subplots(2, 10, figsize=(10, 4))
axes = axes.flatten()

for i in range(n_plots):
    grid = grids[i].reshape(11, 11)
    im = axes[i].imshow(grid, cmap='OrRd', interpolation='none')
    
    axes[i].set_title(f'Grid {i+1}')
    axes[i].axis('off')
    
grids.shape
```

### sample from grids at target distances + fit

```python
N_SUBS = 1  # Fitting x subject
N_BLOCKS = 1  # Single block

add_noise = 4
target_n = len(opts)

model_outputs = {}
model_outputs['rbf'] = {}
model_outputs['lin'] = {}
results_df = []

# Loop over possible distance combinations (reduced)
for n in range(target_n):
    
    target_distances = opts[n]    
    sampled_coords = sample_coordinates_by_distance(distance_euc, target_distances)
    print("Sampled coordinates:", sampled_coords)
    
    # Extract sampled values from the grid based on sampled coordinates
    df = extract_sampled_values(grids, sampled_coords)

    # Add some noise to ratings
    noise = np.random.normal(loc=0, scale=4, size=df['rating'].shape)
    df["rating"] = df["rating"] + noise
    df["rating"] = np.clip(df["rating"], 0, 100)
    df["rating"] = np.round(df["rating"]).astype(int)
    
    # Loop over ls (here ID!)
    data = df
    unique_ids = data["ID"].unique()
    for id_val in unique_ids:
        
        # Subset data for current ID
        subset = data[data["ID"] == id_val]
        N_TRIALS = len(subset["rating"])

        # Initialize arrays for shown and rated tiles
        prompt = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
        rating = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
        csloc = np.zeros((N_SUBS, N_BLOCKS, 1))
        csval = np.zeros((N_SUBS, N_BLOCKS, 1))

        # Fill with data
        prompt[0, 0, :] = subset["promptIDX"].values
        rating[0, 0, :] = subset["rating"].values
        csloc[0, 0, :] = cs  # Assuming `cs` is defined somewhere
        csval[0, 0, :] = val  # Assuming `val` is defined somewhere

        # Reshape data
        prompt_flat = np.reshape(prompt, (N_SUBS * N_BLOCKS, N_TRIALS))
        rating_flat = np.reshape(rating, (N_SUBS * N_BLOCKS, N_TRIALS))
        csloc_flat = np.reshape(csloc, (N_SUBS * N_BLOCKS, 1))
        csval_flat = np.reshape(csval, (N_SUBS * N_BLOCKS, 1))

        # Convert to JAX arrays
        prompt_flat = jax.device_put(prompt_flat.astype(int))
        rating_flat = jax.device_put(rating_flat)
        csloc_flat = jax.device_put(csloc_flat.astype(int))
        csval_flat = jax.device_put(csval_flat.astype(int))

        #fit model to data: RBF
        mod = 'rbf'
        model_output = sampleLUI(
            grid_model_SIM,
            prompt_flat,
            rating_flat,
            csloc_flat,
            csval_flat,
            distances,
            N_SUBS,
            N_BLOCKS,
            N_TRIALS,
            n_samples=3000, #1500,
            n_warmup=500,
            num_chains=3,
        )

        posterior_samples = model_output.get_samples()
        model_output.print_summary()
    
        #save all relevant bits
        waic_data = az.from_numpyro(model_output)
        waic_result = az.waic(waic_data) #higher elpd_waic means a better model fit
        loo_result = az.loo(waic_data)
        print("elpd_WAIC:", waic_result.elpd_waic, "elpd_LOO:", loo_result.elpd_loo)

        results_df.append({
            'distances': target_distances,
            'ID': id_val,
            'EWAIC': waic_result.elpd_waic,
            'LOO': loo_result.elpd_loo,
            'SIM':'rbf' if id_val < 10 else 'lin',
            'FIT': mod,
            'LS': subset["ls_index"].unique(),
            'model_output': model_output,
            'posterior_samples': posterior_samples
            
        })
        
        #fit model to data: LIN
        mod = 'lin'
        model_output = sampleLUI(
            grid_model_LIN,
            prompt_flat,
            rating_flat,
            csloc_flat,
            csval_flat,
            distances,  # Assuming `distances` is defined
            N_SUBS,
            N_BLOCKS,
            N_TRIALS,
            n_samples=3000, #1500,
            n_warmup=500,
            num_chains=3,
        )

        posterior_samples = model_output.get_samples()
        model_output.print_summary()

        #WAIC - assess model fit
        waic_data = az.from_numpyro(model_output)
        waic_result = az.waic(waic_data) #higher elpd_waic means a better model fit
        loo_result = az.loo(waic_data)
        print("elpd_WAIC:", waic_result.elpd_waic, "elpd_LOO:",loo_result.elpd_loo)
    
        results_df.append({
            'distances': target_distances,
            'ID': id_val,
            'EWAIC': waic_result.elpd_waic,
            'LOO': loo_result.elpd_loo,
            'SIM':'rbf' if id_val < 10 else 'lin',
            'FIT': mod,
            'LS': subset["ls_index"].unique(),
            'model_output': model_output,
            'posterior_samples': posterior_samples
        })
        
```

### get bet fitting model 

```python
df_fit = pd.DataFrame(results_df)
df_fit['distances'] = df_fit['distances'].apply(lambda x: tuple(x) if isinstance(x, list) else x)

# Group by 'distances' and 'ID/ls' and find the row with the maximum 'EWAIC'
best_fit_per_id = (
    df_fit.loc[df_fit.groupby(['distances', 'ID'])['EWAIC'].idxmax()]
    .reset_index(drop=True)
    .rename(columns={'FIT': 'best'})
)

# Merge back into the original DataFrame
df_rbf = df_fit.merge(best_fit_per_id[['distances', 'ID', 'best']], on=['distances', 'ID'], how='left')
df_rbf['EWAIC'] = df_rbf['EWAIC'].round(2)

df_rbf
```

### Conf matrix to check recovery

```python
unique_distances = df_rbf['distances'].unique()
cols = 3
rows = (len(unique_distances) + cols - 1) // cols
fig, axes = plt.subplots(rows, cols, figsize=(9, 2 * rows))
axes = axes.flatten()

for i, distance in enumerate(unique_distances): 
    df_subset = df_rbf[df_rbf['distances'] == distance]
    confusion_matrix = pd.crosstab(df_subset['FIT'], df_subset['best'], rownames=['FIT'], colnames=['best'], margins=False)
    
    # Plot
    sns.heatmap(confusion_matrix, annot=True, fmt='d', cmap='OrRd', ax=axes[i], cbar=False)
    
    axes[i].set_title(f" {distance}")
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('True')
```
