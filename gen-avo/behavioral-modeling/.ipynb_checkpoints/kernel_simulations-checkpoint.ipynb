```python
def linear_kernel_directional(X: jnp.ndarray, ls: float, angle: float, Xs: jnp.ndarray = None) -> jnp.ndarray:
    """Compute linear kernel with directional decay based on angle."""
    if Xs is None:
        Xs = X
    
    # Direction vector based on angle
    direction_vector = jnp.array([jnp.cos(angle), jnp.sin(angle)])
    
    # Project points X and Xs onto direction vector
    X_proj = jnp.dot(X, direction_vector)
    Xs_proj = jnp.dot(Xs, direction_vector)

    # Kernel based on projections (dot product)
    K = jnp.outer(X_proj, Xs_proj)
    
    # Scale the kernel by length scale (ls) ** not working 
    K_scaled = K * ls
    return K_scaled


def gp_predict_linear_kernel(cs_loc: jnp.ndarray, cs_val: jnp.ndarray, X: jnp.ndarray, ls: float, angle: float) -> jnp.ndarray:
    """Gaussian process prediction with directional linear kernel."""
    
    # Compute linear kernel for training points (cs_loc)
    K = linear_kernel_directional(cs_loc, ls=ls, angle=angle) + 1e-4 * jnp.eye(cs_loc.shape[0])  # Add small noise term for stability
    
    # Perform Cholesky decomposition of K_train
    L = jax.scipy.linalg.cho_factor(K, lower=True, overwrite_a=False, check_finite=False)
    
    # Solve for the woodbury vector using Cholesky factor
    woodbury_vector = jax.scipy.linalg.cho_solve(L, cs_val[:, None], check_finite=False, overwrite_b=False)
    
    # Compute kernel between training points and test points (cs_loc vs X)
    KK = linear_kernel_directional(cs_loc, ls=ls, angle=angle, Xs=X)
    
    # Compute mean prediction for all test points
    mu = jnp.dot(KK.T, woodbury_vector).ravel()
    #mu = jnp.clip(mu, 0, 100)
    
    # Compute variance for test points
    KKK = linear_kernel_directional(X, ls=ls, angle=angle)
    tmp = jax.scipy.linalg.solve_triangular(L[0], KK, lower=True, check_finite=False, overwrite_b=False)
    var = jnp.diag(KKK) - jnp.sum(tmp ** 2, axis=0)
    
    return mu, var


# Parameters
GRID_SIZE = 11
x1 = np.arange(GRID_SIZE)[:, None]
x2 = np.arange(GRID_SIZE)[:, None]
X = jnp.dstack(np.meshgrid(x1.squeeze(), x2.squeeze())).reshape(-1, 2)  # All coordinates on the grid

cs_loc = jnp.array([[5, 5]])  
cs_val = jnp.array([60])

# Range of length scales (ls) from 0.5 to 10.0 in steps of 0.5
ls_values = np.arange(0.5, 5.5, 0.5)
angle = jnp.radians(120)  # Decay direction


fig, axs = plt.subplots(2, len(ls_values) // 2, figsize=(20, 10))
axs = axs.ravel()

for i, ls in enumerate(ls_values):

    mu, var = gp_predict_linear_kernel(cs_loc, cs_val, X, ls=ls, angle=angle)
    mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)

    # Plot
    ax = axs[i]
    im = ax.imshow(mu_grid, extent=[0, GRID_SIZE-1, 0, GRID_SIZE-1], cmap="coolwarm")
    ax.scatter(cs_loc[:, 0], cs_loc[:, 1], color='black', label="Training Points")
    ax.set_title(f"Length Scale: {ls:.1f}, Angle: {angle}°")
    ax.axis('off')

    fig.colorbar(im, ax=ax)
```

```python
#@jit
def lin_distances(X: jnp.ndarray, Xs: jnp.ndarray, cs: tuple, angle: float) -> jnp.ndarray:
    """Compute projections and distances between two sets of points."""
    cx, cy = cs
    X_x, X_y = X.T
    Xs_x, Xs_y = Xs.T
    
    angle_rad = jnp.deg2rad(angle)
    direction_vector = jnp.array([jnp.cos(angle_rad), jnp.sin(angle_rad)])

    grid_vectors_X = jnp.stack([X_x - cx, X_y - cy], axis=-1)
    grid_vectors_Xs = jnp.stack([Xs_x - cx, Xs_y - cy], axis=-1)

    projections_X = jnp.dot(grid_vectors_X, direction_vector)
    projections_Xs = jnp.dot(grid_vectors_Xs, direction_vector)

    # Calculate distances
    distances = projections_X[:, None] - projections_Xs[None, :]

    return distances

#@jit
def linear_decay_kernel(X: jnp.ndarray, max_value: float, Xs: jnp.ndarray = None, cs: tuple = (0, 0), angle: float = 0, decay: float = 1.0) -> jnp.ndarray:
    if Xs is None:
        Xs = X

    # Get distances depending on angle
    distances = lin_distances(X, Xs, cs, angle)
    #print(distances.shape)

    # Decay on one side (distances < 0), increase on the other side (distances > 0)
    decay_side = max_value - (max_value / decay) * jnp.abs(distances) 
    increase_side = jnp.where(distances > 0, max_value + (max_value / decay) * jnp.abs(distances), 0)  # Increase to max 100 on the other side
    
    # Cap the increasing side at 100
    increase_side = jnp.clip(increase_side, max_value, 100)

    # Combine sides
    K = jnp.where(distances < 0, decay_side, increase_side)
    K = jnp.clip(K, 0, 100) #cap at min and max possible rating 
    
    return K


#@jit
def gp_predict_linear(csloc: jnp.ndarray, val: jnp.ndarray, X: jnp.ndarray, angle: float, decay: float) -> Union[jnp.ndarray, jnp.ndarray]:
    # Create covariance matrix from observed points
    K = linear_decay_kernel(csloc, val, angle=angle, decay=decay) + (1e-4 * jnp.eye(csloc.shape[0])) # Regularization
    L = jnp.linalg.cholesky(K)
    
    # Solve for the woodbury vector
    woodbury_vector = jnp.linalg.solve(L.T, jnp.linalg.solve(L, val))
    
    # Calculate the covariance for the test points
    Kx = linear_decay_kernel(csloc, val, X, angle=angle, decay=decay)

    mu = jnp.dot(Kx.T, woodbury_vector)
    
    # Calculate the full covariance matrix for the new points
    Kxx = linear_decay_kernel(X, val, angle=angle, decay=decay)

    # Variance computation
    tmp = jnp.linalg.solve(L, Kx)
    var = jnp.diag(Kxx) - jnp.sum(jnp.square(tmp), axis=0)
    
    return mu, var


angle = 220 
# Grid initialization
grids_linear = np.zeros((len(lss), GRID_SIZE * GRID_SIZE))

# Plotting
fig, ax = plt.subplots(1, len(lss), figsize=(15, 5))
for i, decay in enumerate(lss):
    mu, var = gp_predict_linear(csloc, jnp.array(val_flat), X, angle=angle, decay=decay)
    mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
    mu_flat = mu.squeeze()
    grids_linear[i, :] = mu_flat  # Collect all grids
    
    im = ax[i].imshow(mu_grid, interpolation='none', cmap=plt.get_cmap('OrRd'), vmin=0)
    ax[i].set_title(f'ls: {decay}')
    ax[i].axis('off')
    ax[i].scatter(cs[0], cs[1], color='black', s=40)
```

### Axis Kernel

```python
#why is @jit not working??
#@jit
def axis_decay_kernel(X: jnp.ndarray, Xs: jnp.ndarray = None, cs: tuple = (0, 0), angle: float = 0, decay: float = 1.0, max_value: float = 80.0) -> np.ndarray:
    """
    Calculates a linear kernel matrix with directional decay based on a specified angle and axis.
    """
    if Xs is None:
        Xs = X

    # start of decay
    cx, cy = cs

    # Reshape X and Xs to separate x and y coordinates
    X_x, X_y = X.T
    Xs_x, Xs_y = Xs.T

    # Convert angle to radians (if provided in degrees)
    angle_rad = jnp.deg2rad(angle)

    # Compute the vector in the direction of the angle
    direction_vector = jnp.array([jnp.cos(angle_rad), jnp.sin(angle_rad)])

    # Create vectors from the center point `cs` to each point in the grid
    grid_vectors_X = jnp.stack([X_x - cx, X_y - cy], axis=-1)
    grid_vectors_Xs = jnp.stack([Xs_x - cx, Xs_y - cy], axis=-1)

    # Project grid vectors onto the direction vector
    projections_X = jnp.dot(grid_vectors_X, direction_vector)
    projections_Xs = jnp.dot(grid_vectors_Xs, direction_vector)

    # Calculate full distance between points and apply decay based on their projection onto the angle direction
    distances = jnp.abs(projections_X[:, None] - projections_Xs[None, :])
    
    decay_values = max_value - (max_value / decay) * distances
    #decay_values = max_value - decay * distances

    # Ensure values are non-negative
    K = jnp.clip(decay_values, 0, max_value)

    return K

#@jit
def gp_predict_axis(X_shown: jnp.ndarray, y_flat: jnp.ndarray, X: jnp.ndarray, angle: float, decay: float, cs: tuple) -> Union[np.ndarray, np.ndarray]:
    """
    Predicts GP mean and variance based on a linear kernel with directional decay.

    Args:
        X_shown (jnp.ndarray): Training input points.
        y_flat (jnp.ndarray): Flattened output values.
        X (jnp.ndarray): Test input points.
        angle (float): Angle in radians that specifies the direction of decay.
        decay (float): Decay factor.

    Returns:
        mu: Predicted mean values.
        var: Predicted variance values.
    """
    cx, cy = cs
    
    # Calculate the kernel matrix K
    K = axis_decay_kernel(X_shown, cs=(cx, cy), angle=angle, decay=decay) + (1e-4 * jnp.eye(X_shown.shape[0]))  # Add jitter for numerical stability

    # Cholesky decomposition
    L = jnp.linalg.cholesky(K)

    # Solve for the "woodbury vector"
    woodbury_vector = jnp.linalg.solve(L.T, jnp.linalg.solve(L, y_flat))

    # Compute the kernel between shown points and test points
    Kx = axis_decay_kernel(X_shown, X, cs=(cx, cy), angle=angle, decay=decay)

    # Compute the mean of the posterior
    mu = jnp.dot(Kx.T, woodbury_vector)

    # Compute the kernel of the test points (auto-covariance)
    Kxx = axis_decay_kernel(X, cs=(cx, cy), angle=angle, decay=decay)  # K(X, X)

    # Solve for the variance
    tmp = jnp.linalg.solve(L, Kx)
    var = jnp.diag(Kxx) - jnp.sum(jnp.square(tmp), axis=0)

    return mu, var


# Parameters and data setup
GRID_SIZE = 11
N_SUBS = 1
N_BLOCKS = 1
N_TRIALS = 1  # Single trial case
y = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))  # Shape: (1, 1, 1)

# Example of shown cs+ value
y[0, 0, 0] = 80  # Set cs+ value for first subject/block
y_flat = np.reshape(y, (N_SUBS * N_BLOCKS,))  

# Example of shown cs+ location
cs = [8, 1]  
X_shown = np.array([cs]) #[x,y]
X_shown = X_shown.astype(np.int32)

# All possible coordinated
x1 = np.arange(GRID_SIZE)[:, None]
x2 = np.arange(GRID_SIZE)[:, None]
X = jnp.dstack(jnp.meshgrid(x1.squeeze(), x2.squeeze())).reshape(-1, 2)  # 11x11 grid

# Define decay factors
decay_factors = np.arange(0.5, 5.5, 0.5)

fig, ax = plt.subplots(1, len(decay_factors), figsize=(15, 5))
for i, decay in enumerate(decay_factors):
    angle = np.random.randint(1, 361)
    
    # Predict GP mean and variance for the fixed coordinate with the current decay factor and angle
    mu, var = gp_predict_axis(jnp.array(X_shown), jnp.array(y_flat), X, angle=angle, decay=decay, cs=cs)

    # Reshape for plotting
    mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
    im = ax[i].imshow(mu_grid, interpolation='none', cmap=plt.get_cmap('OrRd'), vmin=0, vmax=80)

    ax[i].set_title(f'Decay: {decay:.2f}')
    ax[i].axis('off')

plt.suptitle('GP Mean Predictions for Different Decay Factors and Angles')
```

```python
#@jit
def linear_decay_kernel(X: jnp.ndarray, Xs: jnp.ndarray = None, cs: tuple = (0, 0), angle: float = 0, decay: float = 1.0, max_value: float = 80.0) -> np.ndarray:
    if Xs is None:
        Xs = X

    cx, cy = cs
    X_x, X_y = X.T
    Xs_x, Xs_y = Xs.T
    
    angle_rad = jnp.deg2rad(angle)
    direction_vector = jnp.array([jnp.cos(angle_rad), jnp.sin(angle_rad)])

    grid_vectors_X = jnp.stack([X_x - cx, X_y - cy], axis=-1)
    grid_vectors_Xs = jnp.stack([Xs_x - cx, Xs_y - cy], axis=-1)

    projections_X = jnp.dot(grid_vectors_X, direction_vector)
    projections_Xs = jnp.dot(grid_vectors_Xs, direction_vector)

    # Calculate full distance and adjust for linear increase/decrease
    distances = projections_X[:, None] - projections_Xs[None, :]

    # Decay on one side (distances < 0), increase on the other side (distances > 0)
    #decay_side = max_value - decay * jnp.abs(distances)  # Decay from max_value
    decay_side = max_value - (max_value / decay) * jnp.abs(distances) 
    increase_side = jnp.where(distances > 0, max_value + decay * distances, 0)  # Increase to max 100 on the other side
    
    # Cap the increasing side at 100
    increase_side = jnp.clip(increase_side, 80, 100)

    # Combine the decaying and increasing values
    K = jnp.where(distances < 0, decay_side, increase_side)

    # Ensure values stay between 0 and max_value (for the decaying side)
    K = jnp.clip(K, 0, 100)

    return np.array(K)



#@jit
def gp_predict_linear(X_shown: jnp.ndarray, y_flat: jnp.ndarray, X: jnp.ndarray, angle: float, decay: float, cs:tuple) -> Union[np.ndarray, np.ndarray]:
    cx, cy = cs
    K = linear_decay_kernel(X_shown, cs=(cx, cy), angle=angle, decay=decay) + (1e-4 * jnp.eye(X_shown.shape[0]))
    L = jnp.linalg.cholesky(K)
    woodbury_vector = jnp.linalg.solve(L.T, jnp.linalg.solve(L, y_flat))
    Kx = linear_decay_kernel(X_shown, X, cs=(cx, cy), angle=angle, decay=decay)
    mu = jnp.dot(Kx.T, woodbury_vector)
    Kxx = linear_decay_kernel(X, cs=(cx, cy), angle=angle, decay=decay)
    tmp = jnp.linalg.solve(L, Kx)
    var = jnp.diag(Kxx) - jnp.sum(jnp.square(tmp), axis=0)
    return mu, var



X_shown = np.array([cs]).astype(np.int32)

#angle 
angle = np.random.randint(1, 361)
angle = 220 #180

grids_linear = np.zeros((len(lss), 121))

fig, ax = plt.subplots(1, len(lss), figsize=(15, 5))
for i, decay in enumerate(lss):
    #angle = np.random.randint(1, 361)
    mu, var = gp_predict_linear(jnp.array(X_shown), jnp.array(y_flat), X, angle=angle, decay=decay, cs=cs)
    mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
    mu_flat = mu.squeeze()
    grids_linear[i, :] = mu_flat #collect all grids
    #print(mu[index])
    
    im = ax[i].imshow(mu_grid, interpolation='none', cmap=plt.get_cmap('OrRd'), vmin=0, vmax=80)
    ax[i].set_title(f'ls: {decay}')
    ax[i].axis('off')
    ax[i].scatter(cs[0], cs[1], color='black', s=40)
```

```python

#@jit
def lin_distances(X: jnp.ndarray, Xs: jnp.ndarray, cs: tuple, angle: float) -> jnp.ndarray:
    """Compute projections and distances between two sets of points."""
    cx, cy = cs
    X_x, X_y = X.T
    Xs_x, Xs_y = Xs.T
    
    angle_rad = jnp.deg2rad(angle)
    direction_vector = jnp.array([jnp.cos(angle_rad), jnp.sin(angle_rad)])

    grid_vectors_X = jnp.stack([X_x - cx, X_y - cy], axis=-1)
    grid_vectors_Xs = jnp.stack([Xs_x - cx, Xs_y - cy], axis=-1)

    projections_X = jnp.dot(grid_vectors_X, direction_vector)
    projections_Xs = jnp.dot(grid_vectors_Xs, direction_vector)

    # Calculate distances
    distances = projections_X[:, None] - projections_Xs[None, :]

    return distances

#@jit
def linear_decay_kernel(X: jnp.ndarray, max_value: float, Xs: jnp.ndarray = None, cs: tuple = (0, 0), angle: float = 0, decay: float = 1.0) -> jnp.ndarray:
    if Xs is None:
        Xs = X

    # Get distances depending on angle
    distances = lin_distances(X, Xs, cs, angle)

    # Decay on one side (distances < 0), increase on the other side (distances > 0)
    decay_side = max_value - (max_value / decay) * jnp.abs(distances) 
    increase_side = jnp.where(distances > 0, max_value + (max_value / decay) * jnp.abs(distances), 0)  # Increase to max 100 on the other side
    
    # Cap the increasing side at 100
    increase_side = jnp.clip(increase_side, max_value, 100)

    # Combine sides
    K = jnp.where(distances < 0, decay_side, increase_side)
    K = jnp.clip(K, 0, 100) #cap at min and max possible rating 
    
    return K


#@jit
def gp_predict_linear(csloc: jnp.ndarray, val: jnp.ndarray, X: jnp.ndarray, angle: float, decay: float) -> Union[jnp.ndarray, jnp.ndarray]:
    # Create covariance matrix from observed points
    K = linear_decay_kernel(csloc, val, angle=angle, decay=decay) + (1e-4 * jnp.eye(csloc.shape[0])) # Regularization
    L = jnp.linalg.cholesky(K)
    
    # Solve for the woodbury vector
    woodbury_vector = jnp.linalg.solve(L.T, jnp.linalg.solve(L, val))
    
    # Calculate the covariance for the test points
    Kx = linear_decay_kernel(csloc, val, X, angle=angle, decay=decay)

    mu = jnp.dot(Kx.T, woodbury_vector)
    
    # Calculate the full covariance matrix for the new points
    Kxx = linear_decay_kernel(X, val, angle=angle, decay=decay)

    # Variance computation
    tmp = jnp.linalg.solve(L, Kx)
    var = jnp.diag(Kxx) - jnp.sum(jnp.square(tmp), axis=0)
    
    return mu, var



```
