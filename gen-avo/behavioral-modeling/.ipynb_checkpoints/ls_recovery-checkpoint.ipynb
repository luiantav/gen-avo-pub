#### 1/ Simulate some value distributions with different ls

```python
N_SUBS = 1
N_BLOCKS = 1
N_TRIALS = 1 #in this case cs+ shown once 
y = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))#3D

# Define Grid
GRID_SIZE = 11
cs = 60 #cs position idx
val = 80 #cs value
row, col = divmod(cs, 11) #position coordinates
cs_position = [row,col]

# Make cartesian grid
x1 = np.arange(GRID_SIZE)[:, None]
x2 = np.arange(GRID_SIZE)[:, None]
X = jnp.dstack(np.meshgrid(x1.squeeze(), x2.squeeze())).reshape(-1, 2) #all coordinates on grid

# Get distances
distances = jnp.sqrt(square_dist(X, 1)) #precompute distances between values in X

y[0,0,0] = val #cs value
y_flat = np.reshape(y, (N_SUBS * N_BLOCKS, N_TRIALS))

mean = 0
lss = np.arange(0.5, 5.5, 0.5) #range ls parameter
#ls_length = len(lss)
grids = np.zeros((len(lss), 121))

for cs in [cs]: # example locations on grid
    f, ax = plt.subplots(1, len(lss), figsize=(20, 5))

    for i in range(len(lss)):
        ls = lss[i]
        cell = np.array(cs).reshape(1, 1)
        cell.astype(np.int32)
        
        #get mu and var
        mu, var = gp_predict(cell, y_flat, distances, ls, mean)
        mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
        mu_flat = mu.squeeze()
        grids[i, :] = mu_flat #collect all grids
    
        #plot
        ax[i].imshow(mu_grid, interpolation='none', cmap=plt.get_cmap('OrRd'))
        ax[i].set_title(f"ls: {ls}")
```

#### 2/ Sample at "random" locations with equal distances (will be changed once we have better strategy)

```python
#get euc distances to sample at different distances from CS+
distance_euc = np.zeros((GRID_SIZE, GRID_SIZE))
for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        distance_euc[i, j] = np.sqrt((i - cs_position[0]) ** 2 + (j - cs_position[1]) ** 2)

#sample based on these distances (just for now)
target_distances = [1, 2, 3, 4, 5]

# Get n coordinates corresponding to these distances where n is length of target_distances
sampled_coords = sample_coordinates_by_distance(distance_euc, target_distances)
print("Sampled coordinates:", sampled_coords)

#Plot
prompt_grid = np.zeros((GRID_SIZE, GRID_SIZE))
for coord in sampled_coords:
    prompt_grid[coord] = 1
prompt_grid[(cs_position[0], cs_position[1])] =2

fig, ax = plt.subplots(figsize=(2, 2))
plt.imshow(prompt_grid, interpolation='none', cmap=plt.get_cmap('OrRd'))
plt.colorbar()
plt.title("Prompts")
plt.show()
```

```python
# Collect all data
df = extract_sampled_values(grids, sampled_coords)
add_noise = 4 #sd 
# Add some noise to ratings
noise = np.random.normal(loc=0, scale=add_noise, size=df['rating'].shape)
df["rating"] = df["rating"] + noise
df["rating"] = np.clip(df["rating"], 0, 100)
df["rating"] = np.round(df["rating"]).astype(int)
```

#### 3/ Fit model to ratings and check if we can recover ls

```python
import arviz as az
data = df
model_outputs = {}
N_SUBS = 1  #fitting x subject (probably not ideal)
N_BLOCKS = 1  #Single block
    
# Get unique IDs
unique_ids = data["ID"].unique()
for id_val in unique_ids:
    
    # Subset data for current ID
    subset = data[data["ID"] == id_val]
    N_TRIALS = len(subset["rating"])

    # Initialise arrays for shown and rated tiles
    prompt = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
    rating = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
    csloc = np.zeros((N_SUBS, N_BLOCKS, 1))
    csval = np.zeros((N_SUBS, N_BLOCKS, 1))

    # Fill with data
    prompt[0, 0, :] = subset["promptIDX"].values
    rating[0, 0, :] = subset["rating"].values
    csloc[0, 0, :] = cs
    csval[0, 0, :] = val

    # Reshape data
    prompt_flat = np.reshape(prompt, (N_SUBS * N_BLOCKS, N_TRIALS))
    rating_flat = np.reshape(rating, (N_SUBS * N_BLOCKS, N_TRIALS))
    csloc_flat = np.reshape(csloc, (N_SUBS * N_BLOCKS, 1))
    csval_flat = np.reshape(csval, (N_SUBS * N_BLOCKS, 1))
    
    # Convert to JAX arrays
    prompt_flat = jax.device_put(prompt_flat.astype(int))
    rating_flat = jax.device_put(rating_flat)
    csloc_flat = jax.device_put(csloc_flat.astype(int))
    csval_flat = jax.device_put(csval_flat.astype(int))

    # Run model and fit across trials
    model_output = sampleLUI(
        grid_model_SIM,
        prompt_flat,
        rating_flat,
        csloc_flat,
        csval_flat,
        distances,
        N_SUBS,
        N_BLOCKS,
        N_TRIALS,
        n_samples = 1500, #4000,
        n_warmup = 500, #2000,
        num_chains = 3,
    )
    
    posterior_samples = model_output.get_samples()
    model_output.print_summary()

    model_outputs[id_val] = {
        "model_output": model_output,
        "posterior_samples": posterior_samples
    }
    
    az_data = az.from_numpyro(model_output)
    #print(az_data)
    #print(az.summary(az_data))
```

### RBF: Repeat for multiple

```python
# Define Grid
GRID_SIZE = 11
cs = 60 #cs position idx
val = 80 #cs value
row, col = divmod(cs, 11) #position coordinates
cs_position = [row,col]

N_SUBS = 1
N_BLOCKS = 1
N_TRIALS = 1 #in this case cs+ shown once 
y = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))#3D


#get euc distances to sample at different distances from CS+
distance_euc = np.zeros((GRID_SIZE, GRID_SIZE))
for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        distance_euc[i, j] = np.sqrt((i - cs_position[0]) ** 2 + (j - cs_position[1]) ** 2)
```

```python
#generate all possible options at distances
all_combinations = [list(comb) for r in range(2, 6) for comb in itertools.combinations(range(1, 6), r)]
opts = list(map(list, set(map(tuple, all_combinations))))
opts

#len(opts)
```

#### Generate RBF data

```python
#generate some data
# Make cartesian grid
x1 = np.arange(GRID_SIZE)[:, None]
x2 = np.arange(GRID_SIZE)[:, None]
X = jnp.dstack(np.meshgrid(x1.squeeze(), x2.squeeze())).reshape(-1, 2) #all coordinates on grid

# Get distances
distances = jnp.sqrt(square_dist(X, 1)) #precompute distances between values in X

y[0,0,0] = val #cs value
y_flat = np.reshape(y, (N_SUBS * N_BLOCKS, N_TRIALS))

mean = 0
lss = np.arange(0.5, 5.5, 0.5) #range ls parameter
#ls_length = len(lss)
grids = np.zeros((len(lss), 121))

for cs in [cs]: # example locations on grid
    f, ax = plt.subplots(1, len(lss), figsize=(20, 5))

    for i in range(len(lss)):
        ls = lss[i]
        cell = np.array(cs).reshape(1, 1)
        cell.astype(np.int32)
        
        #get mu and var
        mu, var = gp_predict(cell, y_flat, distances, ls, mean)
        mu_grid = mu.reshape(GRID_SIZE, GRID_SIZE)
        mu_flat = mu.squeeze()
        grids[i, :] = mu_flat #collect all grids
    
        #plot
        ax[i].imshow(mu_grid, interpolation='none', cmap=plt.get_cmap('OrRd'))
        ax[i].set_title(f"ls: {ls}")
```

#### Fit to RBF data 

```python
N_SUBS = 1  # Fitting x subject
N_BLOCKS = 1  # Single block

add_noise = 4
target_n = len(opts)
model_outputsRBF = {}

# Loop over entries in target_n
for n in range(target_n):
    
    target_distances = opts[n]
    #target_distances = np.arange(0, entry+1)  # From 0 to entry inclusive
    
    sampled_coords = sample_coordinates_by_distance(distance_euc, target_distances)
    print("Sampled coordinates:", sampled_coords)
    
    # Extract sampled values from the grid based on sampled coordinates
    df = extract_sampled_values(grids, sampled_coords)

    # Add some noise to ratings
    noise = np.random.normal(loc=0, scale=4, size=df['rating'].shape)
    df["rating"] = df["rating"] + noise
    df["rating"] = np.clip(df["rating"], 0, 100)
    df["rating"] = np.round(df["rating"]).astype(int)
    
    # Get unique IDs
    data = df
    unique_ids = data["ID"].unique()
    
    # Initialize dict for current target_n
    model_outputsRBF[n] = {}
    
    for id_val in unique_ids:
        
        # Subset data for current ID
        subset = data[data["ID"] == id_val]
        N_TRIALS = len(subset["rating"])

        # Initialize arrays for shown and rated tiles
        prompt = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
        rating = np.zeros((N_SUBS, N_BLOCKS, N_TRIALS))
        csloc = np.zeros((N_SUBS, N_BLOCKS, 1))
        csval = np.zeros((N_SUBS, N_BLOCKS, 1))

        # Fill with data
        prompt[0, 0, :] = subset["promptIDX"].values
        rating[0, 0, :] = subset["rating"].values
        csloc[0, 0, :] = cs  # Assuming `cs` is defined somewhere
        csval[0, 0, :] = val  # Assuming `val` is defined somewhere

        # Reshape data
        prompt_flat = np.reshape(prompt, (N_SUBS * N_BLOCKS, N_TRIALS))
        rating_flat = np.reshape(rating, (N_SUBS * N_BLOCKS, N_TRIALS))
        csloc_flat = np.reshape(csloc, (N_SUBS * N_BLOCKS, 1))
        csval_flat = np.reshape(csval, (N_SUBS * N_BLOCKS, 1))

        # Convert to JAX arrays
        prompt_flat = jax.device_put(prompt_flat.astype(int))
        rating_flat = jax.device_put(rating_flat)
        csloc_flat = jax.device_put(csloc_flat.astype(int))
        csval_flat = jax.device_put(csval_flat.astype(int))

        # Run model and fit across trials
        model_output = sampleLUI(
            grid_model_SIM,
            prompt_flat,
            rating_flat,
            csloc_flat,
            csval_flat,
            distances,  # Assuming `distances` is defined
            N_SUBS,
            N_BLOCKS,
            N_TRIALS,
            n_samples=1500,
            n_warmup=500,
            num_chains=3,
        )

        posterior_samples = model_output.get_samples()
        model_output.print_summary()

        # Save the model output and posterior samples for the current id and target_n
        model_outputsRBF[n][id_val] = {
            "model_output": model_output,
            "posterior_samples": posterior_samples
        }
        
        # Convert to ArviZ format for diagnostics and summaries
        #az_data = az.from_numpyro(model_output)
        #print(az.summary(az_data))
    
  
```

```python
all_means = []  # This will store [mean_val, id_val, target_n]
print(len(opts))
# Loop over all entries in model_outputs (target_n) and compute the mean for each ID
for n in range(len(opts)):

    for id_val in unique_ids:
        ls = lss[id_val]
        # Extract posterior samples for ls_subject_transformed for the current ID
        pid = model_outputs[n][id_val]["posterior_samples"]  # Posterior samples are already stored
        pidplot = pid['ls_subject_transformed'].flatten()  # Flatten the samples
        
        # Compute the mean of the posterior samples for this combination of ID and target_n
        mean_val = jnp.mean(pidplot)

        # Append the mean, id_val, and target_n to all_means
        all_means.append([mean_val, ls, n])
```

```python
from scipy.stats import pearsonr

all_means_np = np.array(all_means)  # This ensures you are using NumPy arrays
df = pd.DataFrame(all_means_np, columns=['recovered', 'ls', 'n'])

# Create subplots for each unique n (where n is the number of ratings and at the same time distances)
cols = 4  # Number of columns (2 in this case)
rows = (len(opts) + 1) // cols +1

fig, axes = plt.subplots(cols, rows, figsize=(4*rows, 4 * cols), sharey=True)
axes = axes.flatten()  # Flatten in case there are multiple axes

results_df = pd.DataFrame(columns=['seq', 'cor'])

for i in range(len(opts)):

    df_subset = df[df['n'] == i]
    
    # Calculate the Pearson corr
    corr, _ = pearsonr(df_subset['ls'], df_subset['recovered'])
    results_df = results_df.append({'seq': opts[i], 'cor': corr}, ignore_index=True)

    # Plot scatter with regression line on the corresponding subplot
    sns.regplot(data=df_subset, x='ls', y='recovered', scatter_kws={'s': 100}, ax=axes[i], color='blue')
    axes[i].set_title(f'option {opts[i]} \nCorr: {corr:.2f}\nnoise_sd: {4:.2f}')
    axes[i].set_xlim(0, 5)
    axes[i].set_ylim(0, 5)

plt.tight_layout()
```
